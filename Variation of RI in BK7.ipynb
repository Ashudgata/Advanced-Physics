# Variation in Refractive Index with Wavelength in BK7 Glass

## Project Outline 
</br> BK7 is a Borosilicate crown glass manufactured by Schott. It is most commonly used as the glass in designing lens, prisms and other optical instruments. In this project, the variation in the Refractive Index of BK7 is plotted with respect to wavelength, with data in the range 5-33 um.
</br> Data Source- https://refractiveindex.info/?shelf=glass&book=BK7-Schott&page=Lane
</br> The refractive index (n) of a material determines how much light bends when passing through it. Variations in refractive index with wavelength are crucial in designing optical systems.

## Methodology


**Step-1:Data Collection:**
The refractive index data for BK7 was collected from the online database RefractiveIndex.INFO in CSV format.

**Step-2:Data Preprocessing:**
The collected data was loaded into a Pandas DataFrame and checked for missing values

**Step-3:Data Analysis:**
A scatter plot was generated to visualize the relationship between wavelength and refractive index.

**Step-4:Data Visualization**
The results were first plotted using Matplotlib, initially without annotations, and then with annotations marking the maximum and minimum refractive indices.


```python
import pandas as pd
import matplotlib.pyplot as plt

n = pd.read_csv('BK7 n.csv')
missing_values = n.isnull().sum()        
#n.isnull() iterates trhought the dataframe and returns True(1) for every missing value
#sum() then adds up how many times 1 has been found to be missing
print(missing_values)
```

```python
#No Missing Values Found
```

```python
plt.figure(figsize=(16,9))
plt.title('Refractive Index Variation of BK7 with Wavelength')
plt.scatter(n['wl'], n['n'], color='brown', label='Refractive Index', s=3)
plt.xlabel('Wavelength(um)')
plt.ylabel('n')
plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Finding Maximum and Minimum refractive index
max_ri = n[n.n == n.n.max()]
min_ri = n[n.n == n.n.min()]
print(max_ri)
print(min_ri)
```

```python
# Maximum refractive index
max_x = max_ri['wl'].values[0]  # Wavelength for maximum refractive index
max_y = max_ri['n'].values[0]   # Maximum refractive index value

# Minimum refractive index
min_x = min_ri['wl'].values[0]  # Wavelength for minimum refractive index
min_y = min_ri['n'].values[0]   # Minimum refractive index value

plt.figure(figsize=(16,9))
plt.title('Refractive Index Variation of BK7 with Wavelength')
plt.scatter(n['wl'], n['n'], color='brown', label='Refractive Index', s=3)
plt.xlabel('Wavelength(um)')
plt.ylabel('n')
plt.annotate(f'Max_ri={max_y:.2f}',  xy=(max_x, max_y), 
             xytext=(max_x - 2, max_y + 0.2), 
             arrowprops=dict(facecolor='black', shrink=0.05))
plt.annotate(f'Min_ri={min_y:.2f}',  xy=(min_x, min_y), 
             xytext=(min_x + 1, min_y - 0.2), 
             arrowprops=dict(facecolor='black', shrink=0.05))

plt.grid()
plt.legend()
plt.tight_layout()
plt.show()
```
